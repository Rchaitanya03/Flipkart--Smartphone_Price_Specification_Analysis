# Flipkart Smartphone Price and Specification Analysis

**Workflow:** Web scraping → save raw data in `df` → data cleaning → EDA → visualizations → business insights.

This notebook follows the approach in your example: `requests`, `BeautifulSoup`, regular expressions, lists, and a pandas DataFrame. It does **not** use `quote_plus`, `urljoin`, `display`, pandas display settings, Seaborn theme settings, or custom helper functions.

> **Important:** Flipkart's HTML classes can change and the site may return a CAPTCHA/blocked page. Run the scraping section first and inspect the status/output. If no products are extracted, do not interpret the empty DataFrame as evidence that no smartphones are listed. Use this only in line with the website's terms and applicable rules; keep request rates low.

## 1. Import libraries

```python
import requests
import re
import time
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Configure the request and search pages

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/137.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8",
    "Referer": "https://www.google.com/"
}

# Search results for 5G smartphones. Scrape at least 10 pages to increase the sample size.
base_url = "https://www.flipkart.com/search?q=mobile+5g&sid=tyy%2C4io"
number_of_pages = 10
```

## 3. Web scraping

The selectors below begin with the classes used in your example (`RG5Slk`, `hZ3P6w DeU9vF`, and `CMXw7N`) and include a few common alternatives. Because Flipkart may change its page structure, the code prints the HTTP status and number of product titles found on each page.

The notebook scrapes a minimum of 10 search-result pages. Some listings may still be removed during cleaning if their title or price cannot be extracted.

```python
Brand = []
Model = []
Price = []
Color = []
RAM = []
ROM = []
Battery = []
Display = []
Reviews = []

for page_no in range(1, number_of_pages + 1):
    url = base_url + "&page=" + str(page_no)
    response = requests.get(url, headers=headers, timeout=20)
    print("Page:", page_no, "| HTTP status:", response.status_code, "| URL:", url)

    soup = BeautifulSoup(response.text, "html.parser")

    # Title selectors based on the provided example, followed by alternative selectors.
    title_nodes = soup.select("div.RG5Slk")
    if not title_nodes:
        title_nodes = soup.select("div.KzDlHZ")
    if not title_nodes:
        title_nodes = soup.select("a[title]")

    # Price and specification selectors based on the provided example and common alternatives.
    price_nodes = soup.select("div.hZ3P6w.DeU9vF")
    if not price_nodes:
        price_nodes = soup.select("div.Nx9bqj")
    feature_nodes = soup.select("div.CMXw7N")
    if not feature_nodes:
        feature_nodes = soup.select("li._21Ahn-")

    print("Product title elements found:", len(title_nodes))

    # Walk upward from each title until a likely product card containing a price is found.
    for title_node in title_nodes:
        title_text = title_node.get_text(" ", strip=True)
        if not title_text or len(title_text) < 4:
            continue

        card = title_node
        for level in range(5):
            if card.parent is None:
                break
            card = card.parent
            card_text = card.get_text(" ", strip=True)
            card_price = card.select_one("div.hZ3P6w.DeU9vF, div.Nx9bqj, div._30jeq3")
            if card_price is not None and len(card_text) < 1800:
                break

        card_text = card.get_text(" ", strip=True)
        price_node = card.select_one("div.hZ3P6w.DeU9vF, div.Nx9bqj, div._30jeq3")
        if price_node is None:
            # Last-resort price extraction from the product-card text.
            price_match = re.search(r"₹\s?[\d,]+", card_text)
            price_text = price_match.group(0) if price_match else np.nan
        else:
            price_text = price_node.get_text(" ", strip=True)

        feature_texts = [node.get_text(" ", strip=True) for node in card.select("div.CMXw7N, li._21Ahn-")]
        features_text = " | ".join(feature_texts)
        if not features_text:
            features_text = card_text

        price_match = re.search(r"[\d,]+", str(price_text))
        price_value = price_match.group(0) if price_match else np.nan

        ram_match = re.search(r"(\d+)\s*GB\s*RAM", features_text, flags=re.I)
        rom_match = re.search(r"(\d+)\s*GB\s*(?:ROM|Storage)", features_text, flags=re.I)
        battery_match = re.search(r"(\d{3,5})\s*mAh", features_text, flags=re.I)
        display_match = re.search(r"(\d+(?:\.\d+)?)\s*cm", features_text, flags=re.I)
        reviews_match = re.search(r"([\d,]+)\s*(?:Ratings|Reviews)", card_text, flags=re.I)

        # Brand is approximated from the first word of the listing title.
        title_clean = re.sub(r"\s+", " ", title_text).strip()
        brand_value = title_clean.split(" ")[0]
        model_value = title_clean[len(brand_value):].strip(" -") or title_clean

        Brand.append(brand_value)
        Model.append(model_value)
        Price.append(price_value)
        color_match = re.search(r"\(([^,()]+),", title_clean)
        Color.append(color_match.group(1).strip() if color_match else np.nan)
        RAM.append(ram_match.group(1) if ram_match else np.nan)
        ROM.append(rom_match.group(1) if rom_match else np.nan)
        Battery.append(battery_match.group(1) if battery_match else np.nan)
        Display.append(display_match.group(1) if display_match else np.nan)
        Reviews.append(reviews_match.group(1).replace(",", "") if reviews_match else np.nan)

    time.sleep(2)

print("Total product records collected:", len(Brand))
```

## 4. Create the raw DataFrame `df` and save the scraped data

```python
data = {
    "Brand": Brand,
    "Model": Model,
    "Price": Price,
    "Color": Color,
    "RAM_GB": RAM,
    "Storage_GB": ROM,
    "Battery_mAh": Battery,
    "Display_cm": Display,
    "Reviews": Reviews
}

df = pd.DataFrame(data)
print("Raw dataset shape:", df.shape)
df.head(10)
```

```python
df.to_csv("flipkart_smartphone_raw.csv", index=False, encoding="utf-8-sig")
print("Saved: flipkart_smartphone_raw.csv")
```

## 5. Data understanding before cleaning

```python
print("Rows and columns:", df.shape)
print("\nColumn data types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isna().sum())
print("\nDuplicate rows:", df.duplicated().sum())
df.head()
```

## 6. Data cleaning

```python
clean_df = df.copy()

# Remove leading/trailing spaces and standardize text columns.
for column in ["Brand", "Model", "Color"]:
    clean_df[column] = clean_df[column].astype("string").str.strip()
    clean_df[column] = clean_df[column].replace({"": pd.NA, "nan": pd.NA, "None": pd.NA})

# Extract color from text inside parentheses when the scraper has not already captured it.
color_from_model = clean_df["Model"].astype("string").str.extract(r"\(([^,()]+),", expand=False).str.strip()
clean_df["Color"] = clean_df["Color"].fillna(color_from_model)

# Convert numeric fields. Price values may contain commas or currency symbols.
clean_df["Price"] = pd.to_numeric(
    clean_df["Price"].astype("string").str.replace(r"[^0-9.]", "", regex=True),
    errors="coerce"
)

for column in ["RAM_GB", "Storage_GB", "Battery_mAh", "Display_cm", "Reviews"]:
    clean_df[column] = pd.to_numeric(clean_df[column], errors="coerce")

# Remove rows without a usable product title or positive price.
clean_df = clean_df.dropna(subset=["Model", "Price"])
clean_df = clean_df[clean_df["Price"] > 0]

# Remove duplicate listings using brand, model, and price, retaining the first occurrence.
clean_df = clean_df.drop_duplicates(subset=["Brand", "Model", "Price"], keep="first")

# Use nullable integer types for whole-number specifications.
for column in ["RAM_GB", "Storage_GB", "Battery_mAh", "Reviews"]:
    clean_df[column] = clean_df[column].round().astype("Int64")

# Standardize brand capitalization without changing model text.
clean_df["Brand"] = clean_df["Brand"].str.strip().str.title()

clean_df = clean_df.reset_index(drop=True)

print("Raw shape:", df.shape)
print("Cleaned shape:", clean_df.shape)
print("Rows removed:", len(df) - len(clean_df))
clean_df.head(10)
```

```python
print("Missing values after cleaning:")
print(clean_df.isna().sum())
print("\nDuplicate rows after cleaning:", clean_df.duplicated().sum())
print("\nData types after cleaning:")
print(clean_df.dtypes)
```

```python
clean_df.to_csv("flipkart_smartphone_cleaned.csv", index=False, encoding="utf-8-sig")
print("Saved: flipkart_smartphone_cleaned.csv")
```

## 7. Feature engineering

```python
# Price bands are analytical groupings, not Flipkart-defined categories.
clean_df["Price_Band"] = pd.cut(
    clean_df["Price"],
    bins=[0, 10000, 20000, 30000, 50000, np.inf],
    labels=["Under ₹10K", "₹10K–₹20K", "₹20K–₹30K", "₹30K–₹50K", "Above ₹50K"],
    right=True
)

clean_df["Has_RAM_Data"] = clean_df["RAM_GB"].notna()
clean_df["Has_Storage_Data"] = clean_df["Storage_GB"].notna()
clean_df["Has_Battery_Data"] = clean_df["Battery_mAh"].notna()

clean_df.head()
```

## 10. Business insights

The next cell generates **data-driven observations from the scraped dataset**. The statements describe listing prices and listing counts, not actual sales, market share, or Flipkart-wide demand. Treat them as indicative because search ranking, promotions, variants, and incomplete scraping can affect the sample.

```python
if clean_df.empty:
    print("No valid product records are available. Fix the scraper before generating business insights.")
else:
    print("BUSINESS INSIGHTS — BASED ON THIS SCRAPED SAMPLE")
    print("-" * 55)

    print(f"1. The dataset contains {len(clean_df)} unique title-price listings across {clean_df['Brand'].nunique()} brands.")
    print(f"2. The average listed price is ₹{clean_df['Price'].mean():,.0f}; the median is ₹{clean_df['Price'].median():,.0f}.")

    leading_brand = clean_df["Brand"].value_counts().idxmax()
    leading_count = clean_df["Brand"].value_counts().max()
    print(f"3. {leading_brand} has the largest number of listings in this sample ({leading_count}). Listing count does not equal sales or market share.")

    band_counts = clean_df["Price_Band"].value_counts()
    if band_counts.notna().any() and band_counts.sum() > 0:
        leading_band = band_counts.idxmax()
        leading_band_count = band_counts.max()
        print(f"4. The most represented price band is {leading_band} ({leading_band_count} listings). This can help identify where the sampled assortment is concentrated.")

    highest = clean_df.loc[clean_df["Price"].idxmax()]
    lowest = clean_df.loc[clean_df["Price"].idxmin()]
    print(f"5. Highest-priced listing in the sample: {(str(highest['Brand']) + ' ' + str(highest['Model']))} — ₹{highest['Price']:,.0f}.")
    print(f"6. Lowest-priced listing in the sample: {(str(lowest['Brand']) + ' ' + str(lowest['Model']))} — ₹{lowest['Price']:,.0f}.")

    if clean_df["RAM_GB"].notna().sum() > 0:
        ram_summary = clean_df.groupby("RAM_GB")["Price"].median().dropna()
        if len(ram_summary) > 0:
            print("7. Median price by RAM capacity (₹):")
            print(ram_summary.round(0).to_string())

    if clean_df["Storage_GB"].notna().sum() > 0:
        storage_summary = clean_df.groupby("Storage_GB")["Price"].median().dropna()
        if len(storage_summary) > 0:
            print("8. Median price by storage capacity (₹):")
            print(storage_summary.round(0).to_string())

    print(f"9. RAM specification coverage is {clean_df['RAM_GB'].notna().mean()*100:.1f}%; storage coverage is {clean_df['Storage_GB'].notna().mean()*100:.1f}%; battery coverage is {clean_df['Battery_mAh'].notna().mean()*100:.1f}%. Low coverage means specification-based comparisons should be interpreted cautiously.")
```

### Recommended business actions

Use the observed results to test these recommendations rather than treating them as proven facts:

- **Assortment planning:** Compare listing counts and price ranges by brand to identify underrepresented price tiers in this scraped search sample.
- **Price positioning:** Compare each brand's median price with the overall median and with comparable RAM/storage configurations.
- **Value-for-money analysis:** Compare price only among phones with similar RAM, storage, battery, and display specifications; avoid ranking phones on price alone.
- **Content quality:** Improve product-page specification completeness if RAM, storage, or battery values are frequently missing.
- **Promotional planning:** If the dataset is collected repeatedly over time, compare price changes across dates to identify discounts and price volatility. This single snapshot cannot establish price trends.
- **Customer feedback:** Review counts may provide a rough listing-level engagement signal, but they are not reliable evidence of sales volume.
