# Flipkart Smartphone Price and Customer Review Analysis
### Beginner–Intermediate Python Web Scraping Project

**Project flow:** Scrape data → create `df` → clean data → EDA → visualizations → business insights.

This version follows the style of the example notebook you shared: `requests`, `BeautifulSoup`, `re`, `pandas`, and a simple page loop. It avoids advanced scraping frameworks and keeps each step understandable.

**Important:** Flipkart may change its HTML or block automated requests. The CSS selectors in your example (such as `RG5Slk`, `hZ3P6w DeU9vF`, and `CMXw7N`) may no longer work. The scraper below tries to identify product links and parse each product card. If no data is returned, inspect the page HTML or use a permitted dataset; do not invent results or bypass access controls.

## 1. Import the required libraries

```python
import requests
import re
import time
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
sns.set_style("whitegrid")
```

## 2. Set the search URL and headers

```python
# Search smartphones on Flipkart
base_url = "https://www.flipkart.com/search"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/137.0.0.0 Safari/537.36",
    "Accept-Language": "en-IN,en;q=0.9",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "Referer": "https://www.google.com/"
}

# Start small; increase after confirming the first page works
number_of_pages = 3
search_query = "smartphone"
```

## 3. Test whether the page can be opened

```python
params = {"q": search_query, "page": 1}
response = requests.get(base_url, params=params, headers=headers, timeout=20)

print("Status code:", response.status_code)
print("Page URL:", response.url)

if response.status_code == 200:
    print("Page request succeeded. Next, check whether product data is present.")
else:
    print("The request was not successful. Flipkart may be blocking the request or limiting access.")
```

## 4. Create helper functions to extract product information

```python
def get_text(element):
    """Return cleaned text from an HTML element."""
    if element:
        return re.sub(r"\s+", " ", element.get_text(" ", strip=True)).strip()
    return None


def get_number(pattern, text):
    """Find the first value matching a regular expression."""
    match = re.search(pattern, text, flags=re.IGNORECASE)
    if match:
        return match.group(1)
    return None


def extract_price(text):
    """Convert a price such as ₹12,999 into the number 12999."""
    match = re.search(r"₹\s*([\d,]+)", text)
    if match:
        return match.group(1).replace(",", "")
    return None
```

## 5. Scrape product cards into a list

We collect all fields from the same product card. This is safer than creating separate lists for brands, prices, and specifications, because separate lists can have different lengths and become mismatched.

The fields are: brand, model/name, price, rating, review count, RAM, storage, battery, display, colour (when available), and product URL.

```python
all_products = []

for page_number in range(1, number_of_pages + 1):
    params = {"q": search_query, "page": page_number}

    try:
        response = requests.get(
            base_url,
            params=params,
            headers=headers,
            timeout=20
        )
        print("Page", page_number, "status:", response.status_code)

        if response.status_code in [403, 429]:
            print("Access denied or rate limited. Stopping the scrape.")
            break

        response.raise_for_status()
        soup = BeautifulSoup(response.text, "html.parser")

        # Find product links. Flipkart's page structure can change.
        product_links = soup.select('a[href*="/p/"]')
        seen_on_page = set()
        products_on_page = 0

        for link in product_links:
            href = link.get("href", "")
            if not href or "/p/" not in href:
                continue

            product_url = "https://www.flipkart.com" + href.split("?")[0]
            if product_url in seen_on_page:
                continue
            seen_on_page.add(product_url)

            # Move up the HTML tree to find a card containing the product price.
            card = link
            for step in range(5):
                if card.parent is None:
                    break
                parent = card.parent
                parent_text = get_text(parent) or ""
                if "₹" in parent_text and len(parent_text) > 30:
                    card = parent
                else:
                    break

            card_text = get_text(card) or ""
            product_name = link.get("title") or get_text(link)

            # Keep only entries that look like product listings with a price.
            price = extract_price(card_text)
            if not product_name or not price:
                continue

            # Extract fields from the card text when present.
            rating = get_number(r"(?<!\d)([0-5]\.\d)\s*(?:★)?", card_text)
            review_count = get_number(
                r"([\d,]+)\s*(?:Ratings?|Reviews?)", card_text
            )
            ram = get_number(r"(\d+)\s*GB\s*RAM", card_text)
            storage = get_number(r"(\d+)\s*GB\s*(?:ROM|Storage)", card_text)
            battery = get_number(r"(\d{3,5})\s*mAh", card_text)
            display_size = get_number(r"(\d{1,2}(?:\.\d+)?)\s*cm", card_text)

            # The first word is used as a simple brand estimate.
            # Review it manually for names such as "POCO" or multi-word model names.
            brand = product_name.split()[0] if product_name else None

            # Colour is not always shown consistently in search results.
            colour = None
            colour_match = re.search(
                r"\(([^()]*)\)", product_name
            )
            if colour_match:
                colour = colour_match.group(1).split(",")[0].strip()

            all_products.append({
                "Brand": brand,
                "Model": product_name,
                "Price": price,
                "Rating": rating,
                "Review_Count": review_count,
                "Color": colour,
                "RAM": ram,
                "ROM": storage,
                "Battery": battery,
                "Display": display_size,
                "Product_URL": product_url,
                "Listing_Text": card_text
            })
            products_on_page += 1

        print("Products collected from this page:", products_on_page)

        if products_on_page == 0:
            print("No product cards were parsed. The HTML selectors may need updating.")
            break

        time.sleep(2)  # A small delay between page requests

    except requests.RequestException as error:
        print("Request failed:", error)
        break

print("Total product records collected:", len(all_products))
```

## 6. Create the raw DataFrame (`df`) and save it

```python
df = pd.DataFrame(all_products)

print("Raw data shape:", df.shape)
display(df.head())

if not df.empty:
    df.to_csv("flipkart_smartphones_raw.csv", index=False, encoding="utf-8-sig")
    print("Raw data saved as flipkart_smartphones_raw.csv")
else:
    print("The DataFrame is empty. Check the request status and page HTML before continuing.")
```

## 7. Understand the raw data

```python
if not df.empty:
    print("Column names:")
    print(df.columns.tolist())

    print("\nData types and non-null values:")
    df.info()

    print("\nMissing values in each column:")
    display(df.isnull().sum().to_frame("Missing_Count"))

    print("\nDuplicate rows:", df.duplicated().sum())
    print("\nNumber of unique product URLs:", df["Product_URL"].nunique())
```

## 8. Data cleaning

We keep `df` as the raw data and create `clean_df` for the cleaned data. Missing specifications stay missing rather than being replaced with zero, because zero RAM or zero battery would be misleading.

```python
if not df.empty:
    clean_df = df.copy()

    # 1. Remove extra spaces from text columns
    text_columns = ["Brand", "Model", "Color", "Product_URL", "Listing_Text"]
    for column in text_columns:
        clean_df[column] = clean_df[column].astype("string").str.strip()
        clean_df[column] = clean_df[column].replace("", pd.NA)

    # 2. Convert numeric columns to numbers
    numeric_columns = ["Price", "Rating", "Review_Count", "RAM", "ROM", "Battery", "Display"]
    for column in numeric_columns:
        clean_df[column] = pd.to_numeric(
            clean_df[column].astype("string").str.replace(",", "", regex=False),
            errors="coerce"
        )

    # 3. Remove duplicate product URLs
    rows_before = len(clean_df)
    clean_df = clean_df.drop_duplicates(subset="Product_URL", keep="first")
    print("Duplicate product rows removed:", rows_before - len(clean_df))

    # 4. Remove rows without a product name or usable price
    rows_before = len(clean_df)
    clean_df = clean_df.dropna(subset=["Model", "Price"])
    print("Rows removed due to missing model/price:", rows_before - len(clean_df))

    # 5. Treat impossible values as missing
    clean_df.loc[clean_df["Price"] <= 0, "Price"] = np.nan
    clean_df.loc[~clean_df["Rating"].between(0, 5), "Rating"] = np.nan
    for column in ["Review_Count", "RAM", "ROM", "Battery", "Display"]:
        clean_df.loc[clean_df[column] <= 0, column] = np.nan

    # 6. Remove rows if price became invalid
    clean_df = clean_df.dropna(subset=["Price"]).reset_index(drop=True)

    print("Raw data shape:", df.shape)
    print("Cleaned data shape:", clean_df.shape)
    display(clean_df.head())
else:
    clean_df = pd.DataFrame()
    print("No data was scraped, so cleaning cannot continue.")
```

```python
if not clean_df.empty:
    print("Missing values after cleaning:")
    display(clean_df.isna().sum().to_frame("Missing_Count"))

    print("\nNumeric summary:")
    display(clean_df[["Price", "Rating", "Review_Count", "RAM", "ROM", "Battery", "Display"]].describe().T)
```

## 9. Create useful analysis columns

```python
if not clean_df.empty:
    # Create price bands to compare budget, mid-range and premium listings.
    clean_df["Price_Category"] = pd.cut(
        clean_df["Price"],
        bins=[0, 10000, 20000, 40000, np.inf],
        labels=["Budget (<=10k)", "Lower Mid (10k-20k)", "Upper Mid (20k-40k)", "Premium (>40k)"],
        include_lowest=True
    )

    # Discount is not calculated here because search cards may not reliably expose
    # both selling price and original price.
    display(clean_df[["Brand", "Model", "Price", "Price_Category", "Rating", "Review_Count"]].head())
```

## 10. Exploratory Data Analysis (EDA)

```python
if not clean_df.empty:
    print("Total product listings:", len(clean_df))
    print("Unique brands:", clean_df["Brand"].nunique())
    print("Average price: ₹{:,.0f}".format(clean_df["Price"].mean()))
    print("Median price: ₹{:,.0f}".format(clean_df["Price"].median()))
    print("Lowest price: ₹{:,.0f}".format(clean_df["Price"].min()))
    print("Highest price: ₹{:,.0f}".format(clean_df["Price"].max()))

    if clean_df["Rating"].notna().any():
        print("Average listed rating: {:.2f}/5".format(clean_df["Rating"].mean()))
    else:
        print("Rating values were not available in the parsed listings.")

    print("\nListings by brand:")
    display(clean_df["Brand"].value_counts().head(10).to_frame("Listing_Count"))

    print("\nListings by price category:")
    display(clean_df["Price_Category"].value_counts().to_frame("Listing_Count"))

    print("\nAverage price by brand (brands with at least 2 listings):")
    brand_analysis = clean_df.groupby("Brand", dropna=True).agg(
        Listing_Count=("Model", "count"),
        Average_Price=("Price", "mean"),
        Median_Price=("Price", "median"),
        Average_Rating=("Rating", "mean")
    )
    brand_analysis = brand_analysis[brand_analysis["Listing_Count"] >= 2]
    display(brand_analysis.sort_values("Listing_Count", ascending=False).round(2).head(15))
```

## 11. Visualizations

```python
if not clean_df.empty:
    # Chart 1: Price distribution
    plt.figure(figsize=(9, 5))
    sns.histplot(data=clean_df, x="Price", bins=20, kde=True)
    plt.title("Smartphone Price Distribution")
    plt.xlabel("Price (₹)")
    plt.ylabel("Number of listings")
    plt.tight_layout()
    plt.show()

    # Chart 2: Top brands by listing count
    top_brands = clean_df["Brand"].value_counts().head(10).index
    plt.figure(figsize=(9, 5))
    sns.countplot(
        data=clean_df[clean_df["Brand"].isin(top_brands)],
        y="Brand",
        order=clean_df["Brand"].value_counts().head(10).index
    )
    plt.title("Top Brands in Scraped Listings")
    plt.xlabel("Number of listings")
    plt.ylabel("Brand")
    plt.tight_layout()
    plt.show()

    # Chart 3: Price category comparison
    plt.figure(figsize=(9, 5))
    sns.countplot(data=clean_df, x="Price_Category", order=[
        "Budget (<=10k)", "Lower Mid (10k-20k)",
        "Upper Mid (20k-40k)", "Premium (>40k)"
    ])
    plt.title("Smartphone Listings by Price Category")
    plt.xlabel("Price category")
    plt.ylabel("Number of listings")
    plt.xticks(rotation=20, ha="right")
    plt.tight_layout()
    plt.show()

    # Chart 4: Ratings distribution, only if ratings were found
    if clean_df["Rating"].notna().any():
        plt.figure(figsize=(8, 5))
        sns.histplot(data=clean_df, x="Rating", bins=10)
        plt.title("Distribution of Listed Smartphone Ratings")
        plt.xlabel("Rating out of 5")
        plt.ylabel("Number of listings")
        plt.tight_layout()
        plt.show()

    # Chart 5: Price vs rating
    rating_data = clean_df.dropna(subset=["Price", "Rating"])
    if len(rating_data) >= 2:
        plt.figure(figsize=(8, 5))
        sns.scatterplot(data=rating_data, x="Price", y="Rating")
        plt.title("Smartphone Price vs Listed Rating")
        plt.xlabel("Price (₹)")
        plt.ylabel("Rating out of 5")
        plt.tight_layout()
        plt.show()
```

## 12. Business insights from the collected data

Run this section after scraping. The statements are calculated from the current sample; they are not pre-written conclusions. Remember that number of listings is not the same as number of sales or market share.

```python
if not clean_df.empty:
    print("BUSINESS INSIGHTS")
    print("-" * 50)

    # 1. Price positioning
    median_price = clean_df["Price"].median()
    print(f"1. The median listed smartphone price is ₹{median_price:,.0f}.")
    print("   Use this as a reference point when comparing budget and premium listings.")

    # 2. Most frequently listed brand
    brand_counts = clean_df["Brand"].value_counts()
    if not brand_counts.empty:
        print(f"2. The most frequently listed brand in this scrape is {brand_counts.index[0]} "
              f"with {brand_counts.iloc[0]} listings.")
        print("   This measures visibility in the scraped sample, not sales or market share.")

    # 3. Most common price segment
    price_counts = clean_df["Price_Category"].value_counts().dropna()
    if not price_counts.empty:
        print(f"3. The largest price segment in this sample is {price_counts.idxmax()} "
              f"({price_counts.max()} listings).")
        print("   This can help identify which price range is most represented in the search results.")

    # 4. Ratings
    rated_products = clean_df.dropna(subset=["Rating"])
    if not rated_products.empty:
        print(f"4. The average listed rating is {rated_products['Rating'].mean():.2f}/5.")
        print("   Compare ratings with review counts; a high rating with few reviews may be less reliable.")
    else:
        print("4. Ratings were not available reliably, so customer sentiment cannot be summarized.")

    # 5. Highest review count
    review_data = clean_df.dropna(subset=["Review_Count"])
    if not review_data.empty:
        popular = review_data.sort_values("Review_Count", ascending=False).iloc[0]
        print(f"5. The listing with the highest parsed review/ratings count is "
              f"'{popular['Model']}' ({popular['Review_Count']:,.0f}).")
        print("   Review volume can indicate engagement, but it does not prove product quality.")
    else:
        print("5. Review counts were not parsed reliably. Check the raw listing text and page structure.")

    print("\nRECOMMENDATIONS")
    print("- Compare phones within similar price bands and specifications.")
    print("- Check rating and review-count values against a few listings manually.")
    print("- Collect data on another day if you want to compare changes over time.")
    print("- Do not claim sales, revenue, or market share from search listings alone.")
else:
    print("No data was collected. Do not report data-driven business insights until the scrape works.")
```

## 13. Save the cleaned dataset

```python
if not clean_df.empty:
    clean_df.to_csv("flipkart_smartphones_cleaned.csv", index=False, encoding="utf-8-sig")
    print("Cleaned dataset saved as flipkart_smartphones_cleaned.csv")
    print("Final cleaned shape:", clean_df.shape)
else:
    print("No cleaned dataset to save.")
```
