# Flipkart Smartphone Price, Brand, and Rating Analysis

Workflow: web scraping → data cleaning → EDA → visualizations → business analysis → export.

**Important:** Flipkart may block automated requests or change its HTML structure. The selectors in this notebook are common examples, so scraping is not guaranteed to work. Check the site's terms and applicable access rules. Results represent listings collected successfully, not the whole marketplace.

Columns retained: `Brand`, `Model`, `Price`, `Rating`, `Review_Count`, and `Price_Category`. No product URL or full listing-text columns are stored.

## 1. Import libraries

```python
import requests
import re
import time
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Configure the search

```python
base_url = "https://www.flipkart.com/search"
search_query = "smartphone"
number_of_pages = 3

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9"
}
```

## 3. Test one search request

```python
params = {"q": search_query}
response = requests.get(base_url, params=params, headers=headers, timeout=20)

print("Status code:", response.status_code)
print("Page length:", len(response.text))
```

## 4. Parse and inspect the page

```python
soup = BeautifulSoup(response.text, "html.parser")
print("Page title:", soup.title.get_text(strip=True) if soup.title else "No page title found")
```

```python
product_cards = soup.select("div[data-id]")
print("Product containers found:", len(product_cards))
```

```python
if product_cards:
    print(product_cards[0].get_text(" ", strip=True)[:1000])
else:
    print("No product containers found. The page may be blocked or selectors may have changed.")
```

## 5. Scrape smartphone listings

```python
all_products = []

for page in range(1, number_of_pages + 1):
    params = {"q": search_query, "page": page}

    try:
        page_response = requests.get(
            base_url, params=params, headers=headers, timeout=20
        )
    except requests.RequestException as error:
        print(f"Page {page} request failed:", error)
        continue

    print("Scraping page:", page, "| Status:", page_response.status_code)

    if page_response.status_code != 200:
        print("Could not access this page.")
        time.sleep(2)
        continue

    page_soup = BeautifulSoup(page_response.text, "html.parser")
    product_cards = page_soup.select("div[data-id]")

    if not product_cards:
        print("No product containers found. HTML may have changed or request may be blocked.")
        time.sleep(2)
        continue

    for card in product_cards:
        name_element = card.select_one("div.KzDlHZ")
        if name_element is None:
            name_element = card.select_one("a[title]")

        price_element = card.select_one("div.Nx9bqj")
        rating_element = card.select_one("div.XQDdHH")
        review_element = card.select_one("span.Wphh3N")

        product_name = None
        if name_element:
            product_name = name_element.get("title") or name_element.get_text(" ", strip=True)

        price_text = price_element.get_text(" ", strip=True) if price_element else None
        rating_text = rating_element.get_text(" ", strip=True) if rating_element else None
        review_text = review_element.get_text(" ", strip=True) if review_element else None

        if product_name:
            brand = product_name.split()[0]

            price = None
            if price_text:
                price_digits = re.sub(r"[^\d]", "", price_text)
                if price_digits:
                    price = int(price_digits)

            rating = None
            if rating_text:
                rating_match = re.search(r"\d+(?:\.\d+)?", rating_text)
                if rating_match:
                    rating = float(rating_match.group())

            review_count = None
            if review_text:
                review_digits = re.sub(r"[^\d]", "", review_text)
                if review_digits:
                    review_count = int(review_digits)

            all_products.append({
                "Brand": brand,
                "Model": product_name,
                "Price": price,
                "Rating": rating,
                "Review_Count": review_count
            })

    time.sleep(2)

print("Total listings collected:", len(all_products))
```

## 6. Create the raw DataFrame

```python
df = pd.DataFrame(all_products)
print("DataFrame shape:", df.shape)
display(df.head(10))
```

## 7. Check collected data

```python
print("Columns:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
```

```python
if df.empty:
    print("The DataFrame is empty. Fix the scraping step before continuing.")
else:
    print("Listings collected:", len(df))
    print("Unique models:", df["Model"].nunique())
```

### Diagnostic cell — use if no listings were collected

```python
diagnostic_soup = BeautifulSoup(response.text, "html.parser")
print("Test response status:", response.status_code)
print("Page title:", diagnostic_soup.title.get_text(strip=True) if diagnostic_soup.title else "No title")
print("Product containers:", len(diagnostic_soup.select("div[data-id]")))
print("\nText preview:")
print(diagnostic_soup.get_text(" ", strip=True)[:1500])
```

## 8. Remove duplicate listings

```python
if not df.empty:
    df = df.drop_duplicates(subset=["Brand", "Model", "Price"]).reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
display(df.head())
```

## 9. Create a working copy

```python
df_clean = df.copy()
display(df_clean.head())
```

## 10. Check missing values

```python
print("Missing values:")
print(df_clean.isnull().sum())
print("\nMissing-value percentages:")
print((df_clean.isnull().mean() * 100).round(2))
```

## 11. Remove rows without essential information

```python
if not df_clean.empty:
    df_clean = df_clean.dropna(subset=["Brand", "Model", "Price"]).copy()
    df_clean["Price"] = pd.to_numeric(df_clean["Price"], errors="coerce")
    df_clean = df_clean.dropna(subset=["Price"])
    df_clean = df_clean[df_clean["Price"] > 0].reset_index(drop=True)

print("Shape after cleaning:", df_clean.shape)
```

## 12. Clean brand and model text

```python
if not df_clean.empty:
    df_clean["Brand"] = df_clean["Brand"].astype("string").str.strip()
    df_clean["Model"] = df_clean["Model"].astype("string").str.strip()
    df_clean = df_clean[
        df_clean["Brand"].notna() &
        df_clean["Model"].notna() &
        (df_clean["Brand"] != "") &
        (df_clean["Model"] != "")
    ].reset_index(drop=True)

display(df_clean.head())
```

## 13. Set data types

```python
if not df_clean.empty:
    df_clean["Price"] = pd.to_numeric(df_clean["Price"], errors="coerce")
    df_clean["Rating"] = pd.to_numeric(df_clean["Rating"], errors="coerce")
    df_clean["Review_Count"] = pd.to_numeric(df_clean["Review_Count"], errors="coerce")
    df_clean["Review_Count"] = df_clean["Review_Count"].round().astype("Int64")

print(df_clean.dtypes)
```

## 14. Validate ratings and review counts

```python
if not df_clean.empty:
    df_clean.loc[(df_clean["Rating"] < 0) | (df_clean["Rating"] > 5), "Rating"] = np.nan
    df_clean.loc[df_clean["Review_Count"] < 0, "Review_Count"] = pd.NA

print("Rating range:", df_clean["Rating"].min(), "to", df_clean["Rating"].max())
print("Negative prices:", (df_clean["Price"] < 0).sum())
print("Missing ratings:", df_clean["Rating"].isna().sum())
print("Missing review counts:", df_clean["Review_Count"].isna().sum())
```

## 15. Create price categories

```python
if not df_clean.empty:
    df_clean["Price_Category"] = pd.cut(
        df_clean["Price"],
        bins=[0, 10000, 20000, 40000, float("inf")],
        labels=["Budget", "Mid-Range", "Premium", "Flagship"],
        right=True
    )

display(df_clean[["Model", "Price", "Price_Category"]].head() if not df_clean.empty else df_clean)
```

## 16. Final cleaning checks

```python
print("Final dataset shape:", df_clean.shape)
print("Duplicate rows:", df_clean.duplicated().sum())
print("\nMissing values:")
print(df_clean.isnull().sum())
display(df_clean.describe(include="all"))
```

## 17. Overall statistics

```python
if not df_clean.empty:
    print("Total listings:", len(df_clean))
    print("Unique brands:", df_clean["Brand"].nunique())
    print("Average price: ₹", round(df_clean["Price"].mean(), 2))
    print("Median price: ₹", round(df_clean["Price"].median(), 2))
    print("Minimum price: ₹", df_clean["Price"].min())
    print("Maximum price: ₹", df_clean["Price"].max())
    print("Average rating:", round(df_clean["Rating"].mean(), 2))
else:
    print("No data available. Fix scraping before analysis.")
```

## 18. Brand-wise product count

```python
brand_counts = df_clean["Brand"].value_counts().rename_axis("Brand").reset_index(name="Product_Count")
display(brand_counts)
```

## 19. Average price by brand

```python
brand_price = df_clean.groupby("Brand", as_index=False).agg(
    Product_Count=("Model", "count"),
    Average_Price=("Price", "mean"),
    Median_Price=("Price", "median")
)
brand_price["Average_Price"] = brand_price["Average_Price"].round(2)
brand_price["Median_Price"] = brand_price["Median_Price"].round(2)
display(brand_price.sort_values("Average_Price", ascending=False))
```

## 20. Average rating by brand

```python
brand_rating = df_clean.groupby("Brand", as_index=False).agg(
    Average_Rating=("Rating", "mean"),
    Rated_Listings=("Rating", "count")
)
brand_rating["Average_Rating"] = brand_rating["Average_Rating"].round(2)
display(brand_rating.sort_values("Average_Rating", ascending=False))
```

## 21. Price category distribution

```python
price_category_counts = df_clean["Price_Category"].value_counts().reindex(
    ["Budget", "Mid-Range", "Premium", "Flagship"]
)
display(price_category_counts)
```

## 22. Top 10 most expensive listings

```python
top_expensive = df_clean.nlargest(10, "Price")
display(top_expensive[["Brand", "Model", "Price", "Rating"]])
```

## 23. Top 10 highest-rated listings

```python
top_rated = df_clean[
    df_clean["Rating"].notna() &
    (df_clean["Review_Count"].fillna(0) >= 10)
].sort_values(["Rating", "Review_Count"], ascending=[False, False]).head(10)
display(top_rated[["Brand", "Model", "Price", "Rating", "Review_Count"]])
```

## 24. Most-reviewed listings

```python
most_reviewed = df_clean.sort_values("Review_Count", ascending=False, na_position="last").head(10)
display(most_reviewed[["Brand", "Model", "Price", "Rating", "Review_Count"]])
```

## 25. Smartphone price distribution

```python
plt.figure(figsize=(10, 5))
plt.hist(df_clean["Price"].dropna(), bins=20, edgecolor="black")
plt.title("Distribution of Smartphone Prices")
plt.xlabel("Price (₹)")
plt.ylabel("Number of Listings")
plt.tight_layout()
plt.show()
```

## 26. Listings by brand

```python
top_brands = df_clean["Brand"].value_counts().head(10)
plt.figure(figsize=(10, 5))
top_brands.plot(kind="bar")
plt.title("Top 10 Brands by Number of Listings")
plt.xlabel("Brand")
plt.ylabel("Number of Listings")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 27. Average price by brand

```python
top_brands_by_count = df_clean["Brand"].value_counts().head(10).index
average_price_plot = (
    df_clean[df_clean["Brand"].isin(top_brands_by_count)]
    .groupby("Brand")["Price"].mean().sort_values(ascending=False)
)
plt.figure(figsize=(10, 5))
average_price_plot.plot(kind="bar")
plt.title("Average Smartphone Price by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Price (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 28. Rating distribution

```python
plt.figure(figsize=(8, 5))
plt.hist(df_clean["Rating"].dropna(), bins=10, edgecolor="black")
plt.title("Distribution of Smartphone Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Listings")
plt.tight_layout()
plt.show()
```

## 29. Price versus rating

```python
plot_data = df_clean.dropna(subset=["Price", "Rating"])
plt.figure(figsize=(9, 5))
plt.scatter(plot_data["Price"], plot_data["Rating"], alpha=0.6)
plt.title("Smartphone Price vs Rating")
plt.xlabel("Price (₹)")
plt.ylabel("Rating")
plt.tight_layout()
plt.show()
```

## 30. Price category chart

```python
price_category_counts = df_clean["Price_Category"].value_counts().reindex(
    ["Budget", "Mid-Range", "Premium", "Flagship"]
).dropna()
plt.figure(figsize=(8, 5))
price_category_counts.plot(kind="bar")
plt.title("Smartphone Listings by Price Category")
plt.xlabel("Price Category")
plt.ylabel("Number of Listings")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 31. Review count by brand

```python
review_by_brand = (
    df_clean.groupby("Brand")["Review_Count"]
    .sum(min_count=1).sort_values(ascending=False).head(10)
)
plt.figure(figsize=(10, 5))
review_by_brand.plot(kind="bar")
plt.title("Top 10 Brands by Total Recorded Review Count")
plt.xlabel("Brand")
plt.ylabel("Sum of Listing Review Counts")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 32. Brand summary

```python
brand_listing_summary = df_clean.groupby("Brand", as_index=False).agg(
    Listings=("Model", "count"),
    Average_Price=("Price", "mean"),
    Average_Rating=("Rating", "mean")
)
brand_listing_summary["Average_Price"] = brand_listing_summary["Average_Price"].round(2)
brand_listing_summary["Average_Rating"] = brand_listing_summary["Average_Rating"].round(2)
brand_listing_summary = brand_listing_summary.sort_values("Listings", ascending=False)
display(brand_listing_summary.head(10))
```

## 33. Compare price segments

```python
segment_summary = df_clean.groupby("Price_Category", observed=False).agg(
    Listings=("Model", "count"),
    Average_Price=("Price", "mean"),
    Average_Rating=("Rating", "mean")
).reset_index()
segment_summary["Average_Price"] = segment_summary["Average_Price"].round(2)
segment_summary["Average_Rating"] = segment_summary["Average_Rating"].round(2)
display(segment_summary)
```

## 34. Affordable, high-rated listings

```python
median_price = df_clean["Price"].median()
affordable_high_rated = df_clean[
    (df_clean["Price"] <= median_price) & (df_clean["Rating"] >= 4.0)
].sort_values(["Rating", "Price"], ascending=[False, True])
display(affordable_high_rated[["Brand", "Model", "Price", "Rating", "Review_Count"]].head(15))
```

## 35. Price and rating correlation

```python
correlation_data = df_clean[["Price", "Rating"]].dropna()
correlation_value = correlation_data["Price"].corr(correlation_data["Rating"])
print("Correlation between price and rating:", round(correlation_value, 3))
print("Correlation measures association, not causation.")
```

## 36. Market summary

```python
print("SMARTPHONE LISTING SUMMARY")
print("-" * 35)
print("Total listings:", len(df_clean))
print("Unique brands:", df_clean["Brand"].nunique())
print("Average price: ₹", round(df_clean["Price"].mean(), 2))
print("Median price: ₹", round(df_clean["Price"].median(), 2))
print("Average rating:", round(df_clean["Rating"].mean(), 2))
print("\nMost represented brands:")
print(df_clean["Brand"].value_counts().head(5))
print("\nPrice segment distribution:")
print(df_clean["Price_Category"].value_counts())
print("\nHighest average-priced brands:")
print(df_clean.groupby("Brand")["Price"].mean().sort_values(ascending=False).head(5).round(2))
```

```python
df_clean.to_csv("flipkart_smartphone_cleaned.csv", index=False)
print("Saved flipkart_smartphone_cleaned.csv")
```
