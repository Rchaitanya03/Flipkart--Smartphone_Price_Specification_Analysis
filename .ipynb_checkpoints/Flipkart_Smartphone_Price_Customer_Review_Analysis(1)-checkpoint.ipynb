# Flipkart Smartphone Price and Customer Review Analysis

**Workflow:** Web scraping → raw DataFrame (`df`) → data cleaning → EDA → visualizations → business insights.

This notebook is designed as a learning/project template. Flipkart changes its page structure frequently and may block automated requests. The scraper therefore checks for usable product cards, saves the raw data, and stops with a clear message if the page cannot be parsed. **Do not bypass CAPTCHAs, login walls, or access controls.** Review text may not be present on search-result pages; the notebook captures listing-level ratings/review counts when available and includes a separate optional product-review-page section.

Use the collected data in line with Flipkart's terms and applicable laws. Scraped prices, ratings, and review counts are a point-in-time snapshot, not a complete market census.

## 1. Import libraries

```python
import re
import time
import random
import warnings
from urllib.parse import urlencode, urljoin

import requests
import numpy as np
import pandas as pd
from bs4 import BeautifulSoup

import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", 100)
pd.set_option("display.max_colwidth", 120)
sns.set_theme(style="whitegrid")

print("Libraries imported.")
```

## 2. Configure the scrape

```python
# Change these settings if needed. Start with 2–3 pages for testing.
SEARCH_QUERY = "smartphone"
MAX_PAGES = 3
REQUEST_DELAY_SECONDS = (2.0, 4.0)
OUTPUT_RAW = "flipkart_smartphones_raw.csv"

BASE_URL = "https://www.flipkart.com/search"
PARAMS = {"q": SEARCH_QUERY, "otracker": "search", "as-show": "on", "as": "off"}

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0.0.0 Safari/537.36",
    "Accept-Language": "en-IN,en;q=0.9",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
}

session = requests.Session()
session.headers.update(headers)
print(f"Search query: {SEARCH_QUERY!r} | Pages requested: {MAX_PAGES}")
```

## 3. Helper functions for parsing

Flipkart often changes CSS class names, so this uses product links and text patterns rather than relying on one class name alone. A page can still return no products if the website layout changes or a bot-check page is returned.

```python
def clean_text(value):
    if value is None:
        return None
    return re.sub(r"\s+", " ", str(value)).strip() or None

def first_match(pattern, text, flags=re.I):
    match = re.search(pattern, text or "", flags)
    return match.group(1).strip() if match else None

def parse_price(text):
    if not text:
        return np.nan
    # Prefer the first rupee-denominated amount.
    match = re.search(r"₹\s*([\d,]+)", text)
    if match:
        return int(match.group(1).replace(",", ""))
    return np.nan

def parse_card(card, product_url):
    text = clean_text(card.get_text(" ", strip=True)) or ""
    title = clean_text(card.get_text(" ", strip=True))
    # Prefer the linked product title, often exposed as an accessible title.
    product_anchor = card.find("a", href=re.compile(r"/p/"))
    if product_anchor:
        title = clean_text(product_anchor.get("title") or product_anchor.get_text(" ", strip=True)) or title

    # Listing-card price and review metadata
    price = parse_price(text)
    rating = first_match(r"(?<!\d)([0-5]\.\d)\s*(?:★|stars?)?", text)
    review_count_raw = first_match(r"([\d,]+)\s*(?:Ratings?\s*&\s*Reviews?|Ratings?|Reviews?)", text)
    original_price = first_match(r"₹\s*([\d,]+)\s*(?:(?:₹\s*)?[\d,]+\s*off)?", text)
    original_price = int(original_price.replace(",", "")) if original_price else np.nan

    # Common phone specifications, if visible in card text
    ram = first_match(r"(\d{1,2})\s*GB\s*RAM", text)
    storage = first_match(r"(\d{1,4})\s*GB\s*(?:ROM|Storage)", text)
    battery = first_match(r"(\d{3,5})\s*mAh", text)
    display = first_match(r"(\d{1,2}(?:\.\d{1,2})?)\s*cm", text)
    brand = title.split()[0] if title else None

    return {
        "Brand": brand,
        "Product_Name": title,
        "Price_Raw": price,
        "Original_Price_Raw": original_price,
        "Rating_Raw": float(rating) if rating else np.nan,
        "Review_Count_Raw": review_count_raw,
        "RAM_GB_Raw": ram,
        "Storage_GB_Raw": storage,
        "Battery_mAh_Raw": battery,
        "Display_cm_Raw": display,
        "Product_URL": product_url,
        "Listing_Text": text,
    }
```

## 4. Scrape listing pages and create the raw DataFrame (`df`)

```python
records = []
page_errors = []

for page_no in range(1, MAX_PAGES + 1):
    params = {**PARAMS, "page": page_no}
    try:
        response = session.get(BASE_URL, params=params, timeout=25)
        print(f"Page {page_no}: HTTP {response.status_code}")
        if response.status_code in (403, 429):
            page_errors.append(f"Page {page_no}: access denied/rate limited (HTTP {response.status_code})")
            break
        response.raise_for_status()
        soup = BeautifulSoup(response.text, "html.parser")

        # Product links are more stable than generated CSS class names.
        product_links = soup.select('a[href*="/p/"]')
        seen_urls = set()
        page_records = []
        for anchor in product_links:
            href = anchor.get("href", "")
            if not href or "/p/" not in href:
                continue
            product_url = urljoin("https://www.flipkart.com", href.split("?")[0])
            if product_url in seen_urls:
                continue
            seen_urls.add(product_url)

            # Walk upward to find a card-like container with price/title information.
            card = anchor
            for _ in range(5):
                parent = card.parent
                if parent is None:
                    break
                parent_text = clean_text(parent.get_text(" ", strip=True)) or ""
                if "₹" in parent_text and len(parent_text) > 40:
                    card = parent
                else:
                    break

            record = parse_card(card, product_url)
            if record["Product_Name"] and record["Price_Raw"] == record["Price_Raw"]:
                page_records.append(record)

        # Avoid duplicate product URLs across pages
        existing = {r["Product_URL"] for r in records}
        new_records = [r for r in page_records if r["Product_URL"] not in existing]
        records.extend(new_records)
        print(f"  Parsed {len(new_records)} new product cards.")
        if not new_records:
            print("  No new products parsed; the site layout may have changed or returned a bot-check page.")
            if not records:
                page_errors.append("No product cards could be parsed from the returned HTML.")
            break

    except requests.RequestException as exc:
        page_errors.append(f"Page {page_no}: {type(exc).__name__}: {exc}")
        print("  Request failed:", exc)
        break

    time.sleep(random.uniform(*REQUEST_DELAY_SECONDS))

df = pd.DataFrame(records)
print(f"\nRaw dataframe shape: {df.shape}")
display(df.head())
if page_errors:
    print("\nScraping notes:")
    for err in page_errors:
        print("-", err)
if df.empty:
    print("\nNo rows were scraped. Do not continue as if live data was collected. Try again later, check the page manually, or load a permitted dataset.")
```

```python
# Save the raw scrape before cleaning so the original captured values are preserved.
if not df.empty:
    df.to_csv(OUTPUT_RAW, index=False, encoding="utf-8-sig")
    print(f"Saved {len(df)} rows to {OUTPUT_RAW}")
else:
    print("Raw CSV not saved because df is empty.")
```

## 5. Understand the raw data

```python
if not df.empty:
    display(df.info())
    display(df.head(10))
    display(df.isna().sum().sort_values(ascending=False).to_frame("missing_values"))
    print("Exact duplicate rows:", df.duplicated().sum())
    print("Unique product URLs:", df["Product_URL"].nunique())
else:
    print("No scraped rows available. Resolve the scraping issue before running analysis.")
```

## 6. Data cleaning

The raw scrape is retained in `df`. The cleaned version is `clean_df`. Numeric fields are converted safely; missing values are not automatically replaced with zero because zero would imply a real price/rating/specification.

```python
if not df.empty:
    clean_df = df.copy()

    # Standardize text columns
    text_cols = ["Brand", "Product_Name", "Product_URL", "Listing_Text"]
    for col in text_cols:
        if col in clean_df.columns:
            clean_df[col] = clean_df[col].astype("string").str.strip()
            clean_df[col] = clean_df[col].replace({"": pd.NA, "nan": pd.NA, "None": pd.NA})

    # Convert numeric columns
    numeric_source_map = {
        "Price_Raw": "Price_INR",
        "Original_Price_Raw": "Original_Price_INR",
        "Rating_Raw": "Rating",
        "RAM_GB_Raw": "RAM_GB",
        "Storage_GB_Raw": "Storage_GB",
        "Battery_mAh_Raw": "Battery_mAh",
        "Display_cm_Raw": "Display_cm",
    }
    for source, target in numeric_source_map.items():
        if source in clean_df.columns:
            clean_df[target] = pd.to_numeric(clean_df[source], errors="coerce")

    if "Review_Count_Raw" in clean_df.columns:
        clean_df["Review_Count"] = pd.to_numeric(
            clean_df["Review_Count_Raw"].astype("string").str.replace(",", "", regex=False),
            errors="coerce"
        )

    # Remove duplicate URLs and rows without usable product name or price
    before = len(clean_df)
    clean_df = clean_df.drop_duplicates(subset=["Product_URL"], keep="first")
    clean_df = clean_df.dropna(subset=["Product_Name", "Price_INR"])
    print(f"Removed {before - len(clean_df)} duplicate/unusable rows.")

    # Validity checks: flag out-of-range values as missing rather than inventing replacements
    if "Rating" in clean_df:
        clean_df.loc[~clean_df["Rating"].between(0, 5), "Rating"] = np.nan
    if "Price_INR" in clean_df:
        clean_df.loc[clean_df["Price_INR"] <= 0, "Price_INR"] = np.nan
    for col in ["RAM_GB", "Storage_GB", "Battery_mAh", "Display_cm", "Review_Count"]:
        if col in clean_df:
            clean_df.loc[clean_df[col] <= 0, col] = np.nan

    # Derive discount only when original price is larger than selling price.
    if "Original_Price_INR" in clean_df:
        clean_df["Discount_INR"] = clean_df["Original_Price_INR"] - clean_df["Price_INR"]
        clean_df.loc[clean_df["Discount_INR"] <= 0, "Discount_INR"] = np.nan
        clean_df["Discount_Pct"] = (
            clean_df["Discount_INR"] / clean_df["Original_Price_INR"] * 100
        )
        clean_df.loc[~clean_df["Discount_Pct"].between(0, 100), "Discount_Pct"] = np.nan

    # Standardize brand labels
    clean_df["Brand"] = clean_df["Brand"].astype("string").str.strip().str.title()

    print("Cleaned shape:", clean_df.shape)
    display(clean_df.head())
else:
    clean_df = pd.DataFrame()
    print("clean_df is empty because no products were scraped.")
```

```python
if not clean_df.empty:
    print("Missing values after cleaning:")
    display(clean_df.isna().sum().sort_values(ascending=False).to_frame("missing_values"))
    print("Duplicate product URLs:", clean_df["Product_URL"].duplicated().sum())
    display(clean_df[["Price_INR", "Rating", "Review_Count", "RAM_GB", "Storage_GB"]].describe().T)
```

## 8. Exploratory Data Analysis (EDA)

```python
if not clean_df.empty:
    print("Rows:", len(clean_df))
    print("Unique brands:", clean_df["Brand"].nunique())
    print("Price range: ₹{:,.0f} – ₹{:,.0f}".format(clean_df["Price_INR"].min(), clean_df["Price_INR"].max()))
    print("Median price: ₹{:,.0f}".format(clean_df["Price_INR"].median()))
    if clean_df["Rating"].notna().any():
        print("Mean listed rating: {:.2f}/5".format(clean_df["Rating"].mean()))
        print("Products with ratings:", clean_df["Rating"].notna().sum())
    if "Review_Count" in clean_df and clean_df["Review_Count"].notna().any():
        print("Median review count:", clean_df["Review_Count"].median())

    print("\nTop brands by product count:")
    display(clean_df["Brand"].value_counts().head(15).to_frame("Product_Count"))

    print("\nPrice summary by brand (brands with at least 2 scraped products):")
    brand_summary = (clean_df.groupby("Brand")
                     .agg(Product_Count=("Product_Name", "count"),
                          Median_Price=("Price_INR", "median"),
                          Average_Price=("Price_INR", "mean"),
                          Average_Rating=("Rating", "mean"),
                          Median_Review_Count=("Review_Count", "median"))
                     .query("Product_Count >= 2")
                     .sort_values("Product_Count", ascending=False))
    display(brand_summary.head(20).round(2))
```

## 9. Visualizations

```python
if not clean_df.empty:
    # 1. Price distribution
    plt.figure(figsize=(10, 5))
    sns.histplot(data=clean_df, x="Price_INR", bins=20, kde=True)
    plt.title("Distribution of Smartphone Listing Prices")
    plt.xlabel("Price (INR)")
    plt.ylabel("Number of products")
    plt.tight_layout()
    plt.show()

    # 2. Number of scraped products by brand
    top_brands = clean_df["Brand"].value_counts().head(12).index
    plt.figure(figsize=(10, 6))
    sns.countplot(data=clean_df[clean_df["Brand"].isin(top_brands)],
                  y="Brand", order=clean_df["Brand"].value_counts().head(12).index)
    plt.title("Top Brands by Number of Scraped Listings")
    plt.xlabel("Number of listings")
    plt.ylabel("Brand")
    plt.tight_layout()
    plt.show()

    # 3. Price by brand
    brand_order = clean_df.groupby("Brand")["Price_INR"].median().sort_values(ascending=False).head(12).index
    plt.figure(figsize=(12, 6))
    sns.boxplot(data=clean_df[clean_df["Brand"].isin(brand_order)],
                x="Brand", y="Price_INR", order=brand_order)
    plt.title("Smartphone Price Spread by Brand")
    plt.xlabel("Brand")
    plt.ylabel("Price (INR)")
    plt.xticks(rotation=40, ha="right")
    plt.tight_layout()
    plt.show()

    # 4. Rating distribution (only when ratings exist)
    if clean_df["Rating"].notna().any():
        plt.figure(figsize=(8, 5))
        sns.histplot(data=clean_df, x="Rating", bins=np.arange(0, 5.6, 0.2))
        plt.title("Distribution of Listed Customer Ratings")
        plt.xlabel("Rating out of 5")
        plt.ylabel("Number of products")
        plt.tight_layout()
        plt.show()

    # 5. Price vs rating
    rating_data = clean_df.dropna(subset=["Price_INR", "Rating"])
    if len(rating_data) >= 3:
        plt.figure(figsize=(9, 5))
        sns.scatterplot(data=rating_data, x="Price_INR", y="Rating", size="Review_Count",
                        sizes=(25, 250), alpha=0.7, legend=False)
        plt.title("Price vs Listed Rating (bubble size = review count where available)")
        plt.xlabel("Price (INR)")
        plt.ylabel("Rating out of 5")
        plt.tight_layout()
        plt.show()

    # 6. Median price vs average rating by brand
    brand_metrics = (clean_df.groupby("Brand")
                     .agg(Products=("Product_Name", "count"),
                          Median_Price=("Price_INR", "median"),
                          Avg_Rating=("Rating", "mean"))
                     .query("Products >= 2").dropna(subset=["Avg_Rating"]))
    if len(brand_metrics) >= 2:
        plt.figure(figsize=(9, 6))
        sns.scatterplot(data=brand_metrics, x="Median_Price", y="Avg_Rating", size="Products", sizes=(50, 300))
        for brand, row in brand_metrics.iterrows():
            plt.annotate(brand, (row["Median_Price"], row["Avg_Rating"]), fontsize=8, xytext=(4, 3), textcoords="offset points")
        plt.title("Brand-Level Median Price vs Average Rating")
        plt.xlabel("Median price (INR)")
        plt.ylabel("Average rating")
        plt.tight_layout()
        plt.show()
else:
    print("No data available to visualize.")
```

## 10. Customer review text (optional extension)

Search-result pages commonly expose an aggregate rating and review/ratings count, but not individual review text. The following optional cell visits a small number of product pages and looks for a visible reviews link. Flipkart may block these requests or require pagination/login; the code does not bypass such restrictions. Run only if you need review text and your access is permitted.

```python
# Optional: set to True to try a small, polite sample of product pages.
FETCH_REVIEW_PAGES = False
MAX_REVIEW_PRODUCTS = 5

review_records = []
if FETCH_REVIEW_PAGES and not clean_df.empty:
    for _, row in clean_df.head(MAX_REVIEW_PRODUCTS).iterrows():
        try:
            resp = session.get(row["Product_URL"], timeout=25)
            if resp.status_code in (403, 429):
                print("Review-page access denied/rate-limited; stopping.")
                break
            resp.raise_for_status()
            page_soup = BeautifulSoup(resp.text, "html.parser")
            # Extract only text that looks like a review; modern layouts may not expose it in HTML.
            candidates = []
            for node in page_soup.find_all(["div", "p", "span"]):
                txt = clean_text(node.get_text(" ", strip=True))
                if txt and 35 <= len(txt) <= 1200 and not txt.startswith("http"):
                    if any(word in txt.lower() for word in ["good", "great", "bad", "excellent", "product", "battery", "camera", "phone"]):
                        candidates.append(txt)
            # Keep unique snippets, acknowledging these are heuristic candidates, not guaranteed review records.
            for snippet in list(dict.fromkeys(candidates))[:10]:
                review_records.append({
                    "Product_Name": row["Product_Name"],
                    "Product_URL": row["Product_URL"],
                    "Candidate_Review_Text": snippet
                })
            time.sleep(random.uniform(3, 5))
        except requests.RequestException as exc:
            print("Could not fetch product page:", exc)
            continue

reviews_df = pd.DataFrame(review_records)
print("Candidate text snippets found:", len(reviews_df))
display(reviews_df.head(10))
if not reviews_df.empty:
    reviews_df.to_csv("flipkart_review_text_candidates.csv", index=False, encoding="utf-8-sig")
    print("Saved heuristic candidates. Manually validate before treating them as customer reviews.")
else:
    print("No review text extracted. Do not interpret listing ratings as review text.")
```

## 11. Business insights (generated from the actual scrape)

```python
if not clean_df.empty:
    print("BUSINESS INSIGHTS — based only on the products successfully scraped")
    print("=" * 65)

    # Price positioning
    median_price = clean_df["Price_INR"].median()
    mean_price = clean_df["Price_INR"].mean()
    print(f"1. Price positioning: median listed price is ₹{median_price:,.0f}; mean is ₹{mean_price:,.0f}.")
    print("   Business use: use the median as a robust reference price; investigate a large mean–median gap for premium-priced outliers.")

    # Brands by assortment
    brand_counts = clean_df["Brand"].value_counts()
    if len(brand_counts):
        print(f"2. Assortment visibility: {brand_counts.index[0]} has the most listings in this scrape ({brand_counts.iloc[0]}).")
        print("   Business use: listing count indicates representation in this sample, not market share or sales.")

    # Ratings
    rated = clean_df.dropna(subset=["Rating"])
    if not rated.empty:
        best_rated = rated.groupby("Brand").agg(Products=("Rating", "size"), Avg_Rating=("Rating", "mean"))
        best_rated = best_rated[best_rated["Products"] >= 2].sort_values("Avg_Rating", ascending=False)
        print(f"3. Customer sentiment: average listed rating across rated products is {rated['Rating'].mean():.2f}/5.")
        if not best_rated.empty:
            print(f"   Among brands with at least 2 rated listings, highest average is {best_rated.index[0]} ({best_rated.iloc[0]['Avg_Rating']:.2f}/5).")
        print("   Business use: compare ratings alongside review counts; ratings from small samples are less stable.")
    else:
        print("3. Customer sentiment: ratings were not reliably available in the scraped cards; no rating conclusion can be made.")

    # Review counts
    reviewed = clean_df.dropna(subset=["Review_Count"])
    if not reviewed.empty:
        most_reviewed = reviewed.sort_values("Review_Count", ascending=False).iloc[0]
        print(f"4. Review engagement: most review-counted listing in this sample is '{most_reviewed['Product_Name']}' ({most_reviewed['Review_Count']:,.0f}).")
        print("   Business use: high review volume may indicate visibility or maturity, but does not by itself prove better quality.")
    else:
        print("4. Review engagement: review counts were not parsed reliably; inspect the raw Listing_Text and update the parser if needed.")

    # Discounting
    discounted = clean_df.dropna(subset=["Discount_Pct"]) if "Discount_Pct" in clean_df else pd.DataFrame()
    if not discounted.empty:
        print(f"5. Price promotions: {len(discounted)} listings have a usable estimated discount; median observed discount is {discounted['Discount_Pct'].median():.1f}%.")
        print("   Business use: compare discount levels with ratings and review counts before recommending promotion changes.")
    else:
        print("5. Price promotions: original prices/discounts were not consistently available; avoid drawing discount conclusions.")

    print("\nRECOMMENDED NEXT STEPS")
    print("- Scrape more pages only if permitted and stable, then rerun the notebook to check whether patterns persist.")
    print("- Compare like-for-like phone specifications (RAM, storage, battery, display) instead of price alone.")
    print("- Validate rating and review-count parsing against a sample of product pages.")
    print("- Treat this as a listing snapshot. Do not label listing counts as sales, revenue, or market share.")
else:
    print("No live data was captured, so data-driven business insights cannot be calculated.")
```

## 13. Project conclusion and limitations

- The notebook keeps raw and cleaned datasets separate.
- It analyzes listing prices, visible product specifications, aggregate ratings, and review counts when those fields are accessible.
- Individual customer review text is optional and may not be available from listing HTML.
- Scraping may fail because of changing page markup, rate limits, or bot protection. A failed scrape should be reported, not replaced with invented results.
- This is a cross-sectional snapshot: it cannot establish price trends over time, sales volume, customer demographics, or causal relationships.
